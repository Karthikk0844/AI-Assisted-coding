```python
def add_numbers(first_number, second_number):
    return first_number + second_number


result = add_numbers(10, 20)
print(result)

```

```python
def divide_numbers(numerator, denominator):
    try:
        result = numerator / denominator
        return result
    except ZeroDivisionError:
        return "Error: Division by zero is not allowed."
    except TypeError:
        return "Error: Please provide numeric values only."


output = divide_numbers(10, 0)
print(output)
```

```python
def calculate_total(marks):
    return sum(marks)


def calculate_average(total_marks, number_of_subjects):
    return total_marks / number_of_subjects


def calculate_grade(average_marks):
    if average_marks >= 90:
        return "A"
    elif average_marks >= 75:
        return "B"
    elif average_marks >= 60:
        return "C"
    else:
        return "F"


def process_student_marks(marks):
    if not marks:
        print("Error: Marks list is empty.")
        return

    if not all(isinstance(mark, (int, float)) for mark in marks):
        print("Error: All marks must be numeric values.")
        return

    total_marks = calculate_total(marks)
    average_marks = calculate_average(total_marks, len(marks))
    grade = calculate_grade(average_marks)

    print("Total Marks:", total_marks)
    print("Average Marks:", average_marks)
    print("Grade:", grade)


student_marks = [78, 85, 90, 66, 88]
process_student_marks(student_marks)
```

```python
def factorial(n):
    """
    Calculate the factorial of a non-negative integer.

    Parameters:
    n (int): A non-negative integer whose factorial is to be calculated.

    Returns:
    int: Factorial of the given number.

    Raises:
    ValueError: If the input is a negative integer.
    """
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")

    result = 1

    # Multiply result by each integer from 1 to n
    for i in range(1, n + 1):
        result *= i

    return result


print(factorial(5))
```

```python
def is_strong_password(password):
    """
    Validate password strength based on multiple security rules.

    Rules:
    - Minimum length of 8 characters
    - At least one uppercase letter
    - At least one lowercase letter
    - At least one digit
    - At least one special character

    Parameters:
    password (str): Password entered by the user

    Returns:
    bool: True if password is strong, False otherwise
    """
    if len(password) < 8:
        return False

    if not any(char.isupper() for char in password):
        return False

    if not any(char.islower() for char in password):
        return False

    if not any(char.isdigit() for char in password):
        return False

    if not any(char in "!@#$%^&*()-_+=<>?/{}[]|" for char in password):
        return False

    return True


user_password = input("Enter password: ")

if is_strong_password(user_password):
    print("Strong password")
else:
    print("Weak password")
```
